```python
from google.colab import drive
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier

drive.mount('/content/drive')
path_sandbox = '/content/drive/MyDrive/DDoS_Sandbox_Dataset/train_sandbox_v3.csv'

df = pd.read_csv(path_sandbox)
df.columns = df.columns.str.strip()
target_attacks = ['BENIGN', 'Slowloris', 'DoS slowloris']
df_filtered = df[df['Label'].isin(target_attacks)].copy()

X = df_filtered.drop(columns=['Label'])
y = df_filtered['Label']

X = X.select_dtypes(include=[np.number])

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X, y)
feature_importances = pd.Series(rf_model.feature_importances_, index=X.columns)
pd.set_option('display.max_rows', None)
pd.options.display.float_format = '{:.4f}'.format
print(feature_importances.sort_values(ascending=False))
```
